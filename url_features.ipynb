```python
def check_host_type(host):
    import ipaddress
    if not host:
        raise Exception("Invalid or missing host")

    clean_host = host.strip("[]")

    try:
        ipaddress.ip_address(clean_host)
        return "IP Address"
    except ValueError:
        return "Domain"
```
